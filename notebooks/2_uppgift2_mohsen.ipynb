```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("../data/athlete_events.csv")
noc = pd.read_csv("../data/noc_regions.csv")

df.head()
```

```python
sports = ['Athletics', 'Swimming', 'Fencing', 'Football']
sport_df = df[df['Sport'].isin(sports)]
sport_df.head()
```

```python
medals_by_country = (
    sport_df.dropna(subset=['Medal'])
    .groupby(['Sport', 'NOC'])
    .size()
    .reset_index(name='Antal')
)

fig = px.bar(
    medals_by_country,
    x='NOC',
    y='Antal',
    color='Sport',
    title='Medaljfördelning mellan länder per sport'
)
fig.show()
```

```python
fig = px.box(
    sport_df.dropna(subset=['Age']),
    x='Sport',
    y='Age',
    color='Sport',
    title='Åldersfördelning i olika sporter'
)
fig.show()
```

### Sammanfattning

Jag analyserade 4 sporter (Athletice, Swimming, Fencing, Football).
Resultaten visar att Frankrike är starkt i Fencing medan länder som USA dominerar i Athletics och Swimming.
De flesta atleter är mellan 20-30 år gamla, men i vissa sporter som Fencing finns även äldre deltagare.
