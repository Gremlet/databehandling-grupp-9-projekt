```python
import pandas as pd

df = pd.read_csv("../data/athlete_events.csv")
df.head()
```

```python
# a. hur många länder som är med i datan?
antal_lander = df['NOC'].nunique()
print(f"Antal länder i datan: {antal_lander}")
```

```python
# b. vilka länder är med? (förkortningarna räcker)
lander = df['NOC'].unique()
print(lander)
```

```python
# c. vilka sporter är med?
sporter = df['Sport'].unique()
print(f"Antal sportr: {len(sporter)}")
print(sporter)
```

```python
# d. vilka medaljtyper finns det?
medaljer = df['Medal'].dropna().unique()
print(medaljer)
```

```python
# e. ta reda på statistik för åldern: medelvärde, median, min, max, standardavvikelse,
print(df['Age'].describe())
```

```python
# f. utforska datan vidare med egna frågor
import plotly.express as px

fig = px.histogram(df, x='Sex', title='Könsfördelning bland atleter')
fig.show()
```
