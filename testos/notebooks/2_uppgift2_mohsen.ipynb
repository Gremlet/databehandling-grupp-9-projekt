```python
import pandas as pd
import plotly.express as px
```

```python
# Ladda in datafilerna
df = pd.read_csv("../../data/athlete_events.csv")
noc = pd.read_csv("../../data/noc_regions.csv")

df.head()
```

```python
# Slå ihop NOC-informationen för att få landsnamn
df = df.merge(noc, how="left", left_on="NOC", right_on="NOC")
```

```python
# Filtrera fram endast Frankrike 
france = df[df['region'] == 'France'].copy()
```

```python
# Vald Athetics, Swimming, Cycling att analysera
selected_sports = ['Athletics', 'Swimming', 'Cycling']
fr_sports = france[france['Sport'].isin(selected_sports)]
```

```python
# Diagram 1 - Medaljfördelning i sporterna 
# Filtera bort de rader som saknar medalj

medals = fr_sports[fr_sports['Medal'].notna()]

fig1 = px.histogram(
    medals,
    x='Sport',
    color='Medal',
    title='Medaljfördelning för Frankrike i utvalda sporter',
    barmode='group'
)
fig1.show()
```

```python
# Diagram 2 - Åldersfördelning i sporterna
age_data = fr_sports[fr_sports['Age'].notna()]

fig2 = px.histogram(
    age_data,
    x='Age',
    color='Sport',
    nbins=20,
    barmode='group',
    title='Åldersfördelning i sporterna (Frankrike)'
)
fig2.show()
```

```python
# Diagram 3 - Antal medaljer per år
medals_year = medals.groupby(['Year', 'Sport']).size().reset_index(name='Medals')

fig3 = px.line(
    medals_year,
    x='Year',
    y='Medals',
    color='Sport',
    markers=True,
    title='Antal medaljer per år (Frankrike)'
)
fig3.show()
```

```python
# Diagram 4 - Könsfördelning i sporterna
fig4 = px.histogram(
    fr_sports,
    x='Sport',
    color='Sex',
    barmode='group',
    title='Könsfördelning i sporterna (Frankrike)'
)
fig4.show()
```
