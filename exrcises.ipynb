```python
weight_women = os_df.loc[os_df["Sex"] == "F", "Weight"]
weight_men = os_df.loc[os_df["Sex"] == "M", "Weight"]

gold_winners = os_df[os_df["Medal"] == "Gold"]
silver_winners = os_df[os_df["Medal"] == "Silver"]
bronze_winners = os_df[os_df["Medal"] == "Bronze"]


average_gold_weight_women   = gold_winners.loc[gold_winners["Sex"] == "F", "Weight"].mean()
average_silver_weight_women = silver_winners.loc[silver_winners["Sex"] == "F", "Weight"].mean()
average_bronze_weight_women = bronze_winners.loc[bronze_winners["Sex"] == "F", "Weight"].mean()

average_gold_weight_men   = gold_winners.loc[gold_winners["Sex"] == "M", "Weight"].mean()
average_silver_weight_men = silver_winners.loc[silver_winners["Sex"] == "M", "Weight"].mean()
average_bronze_weight_men = bronze_winners.loc[bronze_winners["Sex"] == "M", "Weight"].mean()


# weight_men.mean()
# weight_women.mean()
print(f"Womens gold medalist weight mean is: {average_gold_weight_women:.2f} kg")
print(f"Womens silver medalist weight mean is: {average_silver_weight_women:.2f} kg")
print(f"Womens bronze medalist weight mean is: {average_bronze_weight_women:.2f} kg")
print(f"Mens gold medalist weight mean is: {average_gold_weight_men:.2f} kg")
print(f"Mens silver medalist weight mean is: {average_silver_weight_men:.2f} kg")
print(f"Mens bronze medalist weight mean is: {average_bronze_weight_men:.2f} kg")
print("This proves that the heavier you are the more likely you are to win gold and that it's the same for both sexes!")
```

```python

clean_df = os_df.dropna(subset=["Medal"])


unique_medals = clean_df.drop_duplicates(subset=["NOC", "Team", "Event", "Year", "Medal"])


medal_count = unique_medals["NOC"].value_counts().head(10)


top10_df = medal_count.reset_index()
top10_df.columns = ["NOC", "Unique_Medal_Teams"]


sns.barplot(data=top10_df, x="Unique_Medal_Teams", y="NOC", palette="viridis")
plt.title("Topp 10 länder med flest unika medaljlag")
plt.xlabel("Antal unika lag/individer med medalj")
plt.ylabel("Land (NOC)")
plt.show()
```

```python
france_df = os_df[os_df["NOC"] == "FRA"]
```

```python
import hashlib as hl

fra_anon_df = france_df.copy()

fra_anon_df["Name"] = fra_anon_df["Name"].apply(lambda x:hl.sha256(str(x).encode("utf-8")).hexdigest())



fra_anon_df.head(5)
```

```python
unique_medals = france_df.drop_duplicates(subset=["Sport", "Team", "Event", "Year", "Medal"])


sport_counts = unique_medals.groupby("Sport")["Medal"].count().sort_values(ascending=False).head(10).reset_index()
sport_counts.columns = ["Sport", "Unique_Medal_Count"]


sns.barplot(data=sport_counts, x="Unique_Medal_Count", y="Sport", palette="viridis")
plt.title("Topp 10 sporter där Frankrike tagit flest medaljer")
plt.xlabel("Antal unika medaljer")
plt.ylabel("Sport")
plt.show()
```

```python
unique_medals = france_df.drop_duplicates(subset=["Team", "Event", "Year", "Medal"])

year_counts = unique_medals.groupby("Year")["Medal"].count().reset_index()
year_counts.columns = ["Year", "Unique_Medal_Count"]

# Rita linjediagram
plt.figure(figsize=(10,6))
sns.lineplot(data=year_counts, x="Year", y="Unique_Medal_Count", marker="o")
plt.title("Frankrike – antal unika medaljer per OS-år")
plt.xlabel("År")
plt.ylabel("Antal unika medaljer")
plt.show()
```
